```python
# ==========================================
# INSTALL LIBRARIES
# ==========================================
!pip install -q openai-whisper transformers gtts
!apt-get -qq install ffmpeg

# ==========================================
# IMPORTS
# ==========================================
from google.colab import files
import whisper
from transformers import pipeline
from gtts import gTTS
from IPython.display import Audio, display

# ==========================================
# UPLOAD AUDIO FILE
# ==========================================
print("Upload your audio file (.wav/.mp3)")
uploaded = files.upload()

audio_file = list(uploaded.keys())[0]

print("Audio Loaded:", audio_file)

# ==========================================
# LOAD WHISPER MODEL
# ==========================================
print("\nLoading Whisper Model...")
model = whisper.load_model("base")

# ==========================================
# SPEECH TO TEXT
# ==========================================
print("\nConverting Audio to Text...")

result = model.transcribe(audio_file)

text = result["text"]

print("\nRecognized Text:")
print(text)

# ==========================================
# SENTIMENT ANALYSIS
# ==========================================
print("\nAnalyzing Sentiment...")

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

sentiment_result = classifier(text)

label = sentiment_result[0]["label"]
score = sentiment_result[0]["score"]

print("\nSentiment:", label)
print("Confidence:", round(score, 4))

# ==========================================
# FINAL POSITIVE / NEGATIVE OUTPUT
# ==========================================
if label == "POSITIVE":
    final_output = "Positive"
else:
    final_output = "Negative"

print("\nFinal Output:")
print(final_output)

# ==========================================
# TEXT TO SPEECH
# ==========================================
tts = gTTS(final_output)

output_file = "sentiment_result.mp3"

tts.save(output_file)

print("\nPlaying Output Audio...")

display(Audio(output_file, autoplay=True))
```
